# Term Distribution Analysis - eng_latn_300mb

Exploring the distribution of various terms in the corpus, including spammy/suspicious tokens found in the longest tokens list.


```python
from script_bpe.corpus import load_corpus_by_name
from script_bpe.pretokenize import get_pretokenizer
from collections import Counter
import regex as re
from tabulate import tabulate
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from script_bpe.pretokenize import PRETOKENIZER_REGISTRY, get_pretokenizer, ScriptPretokenizerConfig, ScriptEncodingV2
```

```python
# Load corpus
corpus_name = "eng_latn_300mb"
pretokenizer = get_pretokenizer("scriptenc2_cbi")

line_regex = r"[\r\n]*[^\r\n]*" # This pattern will be used by our custom pretokenizer.

PRETOKENIZER_REGISTRY["scriptenc_line_regex"] = ScriptPretokenizerConfig(
    regex_pattern=line_regex,
    script_split=False,
    script_config=ScriptEncodingV2,
)
analysis_pretokenizer = get_pretokenizer("scriptenc_line_regex")
corpus = load_corpus_by_name(corpus_name, analysis_pretokenizer)
print(f"Loaded corpus: {corpus_name}")
```

```python
texts = {pretokenizer.decode(pretoken_seq): freq for pretoken_seq, freq in corpus}
```

```python
spam= [
    "red test testosterone booster",
    "penis enlargement product",
    "male enhancement stamina product",
    "Male Enhancement Formula Male Extra",
]

penis = []
for term in spam:
    for text, freq in texts.items():
        if term in text:
            count = text.count(term)
            penis.append(dict(term=term, freq=freq, count=count, len=len(text), text=text) )

```

```python
pd.DataFrame(penis)
```
